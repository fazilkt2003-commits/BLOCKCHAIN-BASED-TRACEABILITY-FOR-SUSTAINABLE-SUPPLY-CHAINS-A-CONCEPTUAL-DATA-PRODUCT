```python
# Q24. What is the best genre for a debut director? (Highest avg rating for 1st time directors)
h= df.assign(directors_name= df["directors_name"].str.split(",")).explode("directors_name")

s = h.groupby("directors_name")["id"].count()

s = s[s==1].index

debut_df = h[h["directors_name"].isin(s)]

kj = debut_df.assign(genre=debut_df["genre"].str.split(",")).explode("genre")

kj.groupby("genre")["rating"].mean().sort_values(ascending=False).head()
```

```python
# Q25. Do movies with more actors (ensemble cast) earn more?
s= df.assign(stars_name=df["stars_name"].str.split(","))
s["cast_size"]= s["stars_name"].apply(len)
s.groupby("cast_size")["gross_income"].mean().sort_values(ascending=False)
m = s[["cast_size","gross_income"]]
```

```python
s[["cast_size","gross_income"]].corr()
```

```python
# Q26. Which Director has the highest "Hit Rate"? (Percentage of movies rated > 7.5)
m = df.assign(directors_name = df["directors_name"].str.split(",")).explode("directors_name")
total = m.groupby("directors_name")["rating"].count()

hits = m[m["rating"]> 7.5]
hits = hits.groupby("directors_name")["rating"].count()
hit_percent = (hits/total* 100).sort_values(ascending=False)
hit_percent
```
